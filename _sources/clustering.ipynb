### A. Penentuan Jumlah Cluster Menggunakan Elbow Method

Untuk menentukan jumlah cluster yang akan digunakan dalam proses K-Means, dilakukan evaluasi menggunakan Elbow Method. Metode ini menguji beberapa nilai jumlah cluster dan menghitung nilai inertia yang dihasilkan oleh masing-masing model.

Pengujian dilakukan dengan mencoba jumlah cluster mulai dari $K=2$ hingga $K=10$. Setiap nilai $K$ menghasilkan nilai inertia yang berbeda. Inertia menunjukkan total jarak kuadrat antara setiap data dengan centroid dari cluster tempat data tersebut berada.



**NO2**


```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

from google.colab import files
uploaded = files.upload()

nama_file = list(uploaded.keys())[0]
df = pd.read_csv(nama_file)

fitur = df.drop(['id', 'nama', 'daerah'], axis=1)

scaler = StandardScaler()
fitur_scaled = scaler.fit_transform(fitur)

fitur_pca = PCA(n_components=37).fit_transform(fitur_scaled)

inertia = []
K_range = range(2, 11)

for k in K_range:
    kmeans_temp = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    kmeans_temp.fit(fitur_pca)
    inertia.append(kmeans_temp.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(K_range, inertia, marker='o', linestyle='--')
plt.title('Evaluasi Jumlah Cluster dengan Elbow Method')
plt.xlabel('Jumlah Cluster (K)')
plt.ylabel('Inersia (Jarak Kuadrat)')
plt.grid(True)
plt.show()
```

### B. Visualisasi Scatter Plot PCA dan Profiling

Setelah jumlah cluster yang akan digunakan ditentukan melalui Elbow Method, tahap berikutnya adalah menerapkan algoritma K-Means menggunakan jumlah cluster tersebut. Hasil pengelompokan kemudian divisualisasikan menggunakan Scatter Plot PCA.

Visualisasi ini menggunakan PCA 1 sebagai sumbu X dan PCA 2 sebagai sumbu Y. Dengan demikian, persebaran data pada dua komponen utama dapat diamati untuk melihat pola dan kedekatan antarcluster.

Selain visualisasi, dilakukan pula profiling cluster untuk mengetahui karakteristik masing-masing kelompok berdasarkan fitur yang digunakan dalam proses clustering. Hasil profiling dapat digunakan untuk memahami perbedaan karakteristik data pada setiap cluster.



**NO2**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from google.colab import files

# 1. Memilih file hasil ekstraksi fitur
uploaded = files.upload()
nama_file = list(uploaded.keys())[0]

df = pd.read_csv(nama_file)

# Memisahkan identitas dan fitur
identitas = df[['id', 'nama', 'daerah']]
fitur = df.drop(['id', 'nama', 'daerah'], axis=1)

# 2. Standardisasi dan PCA
scaler = StandardScaler()
fitur_scaled = scaler.fit_transform(fitur)

pca = PCA(n_components=37)
fitur_pca = pca.fit_transform(fitur_scaled)

# 3. K-Means
kmeans = KMeans(
    n_clusters=6,
    random_state=42,
    n_init=10
)

clusters = kmeans.fit_predict(fitur_pca)

# Menyimpan hasil cluster
df_hasil = identitas.copy()
df_hasil['Cluster'] = clusters

# 4. Visualisasi Scatter Plot
plt.figure(figsize=(15, 7))

sns.scatterplot(
    data=df_hasil,
    x='daerah',
    y='Cluster',
    hue='Cluster',
    palette='tab10',
    s=100
)

plt.title('Scatter Plot Persebaran Daerah Berdasarkan Cluster')
plt.xlabel('Nama Daerah')
plt.ylabel('Cluster')
plt.xticks(rotation=90)
plt.legend(title='Cluster')
plt.grid(True)
plt.show()

# 5. Profiling cluster berdasarkan daerah
profil_daerah = (
    df_hasil
    .groupby(['Cluster', 'daerah'])
    .size()
    .reset_index(name='Jumlah')
)

top_3_per_cluster = (
    profil_daerah
    .sort_values(
        by=['Cluster', 'Jumlah'],
        ascending=[True, False]
    )
    .groupby('Cluster')
    .head(3)
)

print('Profil 3 daerah teratas pada setiap cluster:')
print(top_3_per_cluster)
```
